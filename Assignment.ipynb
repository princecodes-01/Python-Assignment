```python
import random

numbers = [random.randint(1, 100) for i in range(10)]

print("Random Numbers:", numbers)

print("Even Numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "Python is easy to learn"

print("Original String:", text)

print("Reversed String:", text[::-1])

print("Reversed Words:", " ".join(text.split()[::-1]))
```
