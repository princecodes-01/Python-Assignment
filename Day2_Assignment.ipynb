```python
import matplotlib.pyplot as plt
```

```python
df["Rating"].value_counts().plot(kind="bar")

plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
df["Rating"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Rating Distribution")
plt.ylabel("")
plt.show()
```

```python
df["Rating"].plot(kind="hist")

plt.title("Rating Histogram")
plt.xlabel("Rating")
plt.show()
```
